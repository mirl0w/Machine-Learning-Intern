## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


A fixed rule like "stale AND visible" (what I built in Notebook 02) can only combine
a small number of conditions a human can hand-pick. It can't learn nonlinear
interactions between signals — for example, that staleness matters more when CTR is
also weak, but less when position is very strong. My own Notebook 02 results showed
this directly: the hand rule scored Precision@50 = 0.600, but a depth-3 decision tree
reached 0.720, purely by finding a slightly better combination of the same signals.
A model can also be retrained as patterns shift over time, while a hand rule stays
frozen until someone manually rewrites it.

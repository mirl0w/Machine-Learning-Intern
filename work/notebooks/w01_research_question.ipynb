## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: which pages should a content reviewer look at first, out of thousands, given
limited review capacity.

Action: a content team member manually reviews the top-ranked pages and decides whether
to refresh, rewrite, or leave them.

Cost of a wrong call:
- False positive (flagged as needing review, but it's fine) → wastes reviewer time,
  the smaller cost.
- False negative (a genuinely declining page never gets flagged) → the page keeps
  losing visibility/traffic silently, the bigger cost, since no one is looking at it.

Because false negatives are costlier than false positives here, precision at a
realistic K (the number of pages a team can actually review) is the right metric —
not raw accuracy.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd, numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_rate = (df["trend_direction"].str.lower() == "down").mean()
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
low_ctr_visible = ((df["impressions_90d"] >= 500) & (df["avg_position"] <= 20) & (df["ctr"] < 0.5)).sum()

print(f"Share of pages currently declining: {declining_rate:.1%}")
print(f"Pages that are stale AND still visible (review candidates): {stale_visible:,} of {len(df):,}")
print(f"Pages with high visibility but weak CTR: {low_ctr_visible:,} of {len(df):,}")
```
